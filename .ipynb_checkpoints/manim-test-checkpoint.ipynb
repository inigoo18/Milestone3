```python
from manim import *

class MathFormula(Scene):
    def construct(self):
        # Create a title
        title = Text("Solving a Quadratic Equation").scale(0.7).to_edge(UP)

        # First formula: general quadratic form
        formula1 = MathTex("x^2", "+", "5x", "+", "6", "=", "0")

        # Second formula: factored form
        formula2 = MathTex("(", "x", "+", "2", ")", "(", "x", "+", "3", ")", "=", "0")

        # Position the formulas
        formula1.next_to(title, DOWN)
        formula2.next_to(formula1, DOWN, buff=1)

        # Animate the title and first formula
        self.play(Write(title))
        self.play(Write(formula1))
        self.wait(1)

        # Transform to the factored form
        self.play(TransformMatchingTex(formula1, formula2))
        self.wait(1)

        # Fade out
        self.play(FadeOut(title), FadeOut(formula2))
```
